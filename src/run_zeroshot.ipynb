```python
# @title
import re
import json

def parse_ruleset(raw_text, ruleset_id="RS-1"):
    lines = raw_text.strip().splitlines()
    ruleset_title = lines[0].rstrip(":").strip()
    rule_lines = [line.strip() for line in lines[1:] if line.strip()]

    rules = []
    rule_id_counter = 1

    for line in rule_lines:
        match = re.match(r"^\* (.*?):\s*(.*?)\s*\[(.*?)\]$", line)
        if match:
            title, definition, source = match.groups()
        else:
            source_match = re.search(r"\[(.*?)\]$", line)
            source = source_match.group(1) if source_match else "Unknown"
            definition = re.sub(r"^\*\s*", "", line)
            definition = re.sub(r"\s*\[.*?\]$", "", definition).strip()
            title = "General rule"

        rules.append({
            "ID": f"{ruleset_id}.{rule_id_counter}",
            "Title": title.strip(),
            "Definition": definition.strip(),
            "Source": source.strip()
        })
        rule_id_counter += 1

    ruleset = {
        "Rule Set": {
            "ID": ruleset_id,
            "Title": ruleset_title,
            "Rules": rules
        }
    }
    return ruleset

def json_to_ruleset(ruleset_json, indent_spaces = 2):
    ruleset = ruleset_json["Rule Set"]
    rule_lines = [f"Rule Set: {ruleset['Title']} (ID: {ruleset['ID']})"]
    indent = " " * indent_spaces

    for rule in ruleset["Rules"]:
        rule_lines.append(f"{indent}Rule ID: {rule['ID']}")
        rule_lines.append(f"{indent*indent_spaces}Title: {rule['Title']}")
        rule_lines.append(f"{indent*indent_spaces}Definition: {rule['Definition']}")
        rule_lines.append(f"{indent*indent_spaces}Source: {rule['Source']}")

    return "\n".join(rule_lines)

def get_template(template_id, rule_key, rule_template='coded_rules'):

    temp = templates.get(template_id)
    instruction_text = temp.get('instruction_text')
    augmented_context = temp.get('augmented_context')

    if template_id == 'r4':
        augmented_context = augmented_context.format(metarules='\n\nMeta-Rules: '+metarules[rule_key])

    response_format = temp.get('response_format')

    if rule_template == 'bulletpoints':

        if isinstance(rule_key, list):
            ruleset_text = "Rule Set: \n\n"
            for i, r in enumerate(rule_key, start=1):
                safety_rules = custom_rules[r].strip()
                rule_name = safety_rules.split(':\n')[0].strip()
                rule_description = safety_rules[len(safety_rules.split(':\n')[0])+1:]
                ruleset_text += rule_description
                ruleset_text += "\n\n"

        elif rule_key == 'all':
            ruleset_text = "Rule Set: \n\n"
            for i, (k, v) in enumerate(custom_rules.items(), start = 1):
                safety_rules = v.strip()
                rule_name = safety_rules.split(':\n')[0].strip()
                rule_description = safety_rules[len(safety_rules.split(':\n')[0])+1:]
                ruleset_text += rule_description
                ruleset_text += "\n\n"
        else:
            safety_rules = custom_rules[rule_key].strip()
            rule_name = safety_rules.split(':\n')[0].strip()
            rule_description = safety_rules[len(safety_rules.split(':\n')[0])+1:]
            ruleset_text = f'Rule Set: {rule_description}"'

    elif rule_template == 'bulletpoints_with_header':

        if isinstance(rule_key, list):
          ruleset_text = "Rule Set: \n\n"
          for i, r in enumerate(rule_key, start=1):
              safety_rules = custom_rules[r].strip()
              rule_name = safety_rules.split(':\n')[0].strip()
              rule_description = safety_rules[len(safety_rules.split(':\n')[0])+1:]
              ruleset_text += f'{i}. Rule: "{rule_name}" \nRelevant Sub-Rules: {rule_description}"'
              ruleset_text += "\n\n"

        elif rule_key == 'all':
            ruleset_text = "Rule Set: \n\n"
            for i, (k, v) in enumerate(custom_rules.items(), start = 1):
                safety_rules = v.strip()
                rule_name = safety_rules.split(':\n')[0].strip()
                rule_description = safety_rules[len(safety_rules.split(':\n')[0])+1:]
                ruleset_text += f'{i}. Rule: "{rule_name}" \nRelevant Sub-Rules: {rule_description}"'
                ruleset_text += "\n\n"
        else:
            safety_rules = custom_rules[rule_key].strip()
            rule_name = safety_rules.split(':\n')[0].strip()
            rule_description = safety_rules[len(safety_rules.split(':\n')[0])+1:]
            ruleset_text = f'Rule Set: "{rule_name}" \nRelevant Sub-Rules: {rule_description}"'

    elif rule_template == 'coded_rules':

        if isinstance(rule_key, list):
            ruleset_text = []
            for i, r in enumerate(rule_key, start=1):
                safety_rules = custom_rules[r].strip()
                ruleset_json = parse_ruleset(safety_rules, ruleset_id=f"RS-{i}")
                ruleset_text.append(json_to_ruleset(ruleset_json))

            ruleset_text = '\n\n'.join(ruleset_text).strip()

        elif rule_key == 'all':
            ruleset_text = []
            for i, (k, v) in enumerate(custom_rules.items(), start = 1):
                safety_rules = v.strip()
                ruleset_json = parse_ruleset(safety_rules, ruleset_id=f"RS-{i}")
                ruleset_text.append(json_to_ruleset(ruleset_json))

            ruleset_text = '\n\n'.join(ruleset_text).strip()

        else:
            safety_rules = custom_rules[rule_key].strip()
            ruleset_json = parse_ruleset(safety_rules, ruleset_id="RS-1")
            ruleset_text = json_to_ruleset(ruleset_json).strip()


    return META_PROMPT_TEMPLATE.format(instruction_text=instruction_text, safety_rules=ruleset_text, augmented_context=augmented_context, response_format=response_format)


def load_image(image_path):
    img = Image.open(image_path)
    if img.mode != 'RGB':
        img = img.convert('RGB')
    return img

def flatten_dict(d):
    flat = {}

    def _flatten(obj):
        if isinstance(obj, dict):
            for k, v in obj.items():
                if isinstance(v, dict) or isinstance(v, list):
                    _flatten(v)
                else:
                    flat[k] = v
        elif isinstance(obj, list):
            for item in obj:
                _flatten(item)

    _flatten(d)
    return flat

def extract_json_prediction(raw_pred):
    try:
        prediction = flatten_dict(ast.literal_eval(repair_json(raw_pred)))

    except (ValueError, SyntaxError):
        return 'Unknown', '', -1

    classification = prediction.get('classification', prediction.get('initial_classification', 'Unknown'))
    try:
        reasoning = prediction.get('reasoning', ast.literal_eval(repair_json(raw_pred))['reasoning'])
    except:
        reasoning = ''
    explanation = prediction.get('explanation', '')
    confidence = float(prediction.get('confidence', -1))

    classification = sanitize_pred_label(classification)

    return classification, reasoning, explanation, confidence

def extract_xml_prediction(raw_pred):

    matching = re.search(r"<classification>\s*([^<]*?)\s*(?:</classification>|(?=<))", str(raw_pred), re.DOTALL)
    matching_ = re.search(r"<CONCLUSION>\s*([^<]*?)\s*(?:</CONCLUSION>|(?=<))", str(raw_pred), re.DOTALL)
    matching__ = re.search(r"<initial_classification>\s*([^<]*?)\s*(?:</initial_classification>|(?=<))", str(raw_pred), re.DOTALL)

    if matching:
        classification = matching.group(1).strip()
    elif matching_:
        classification = matching_.group(1).strip()
    elif matching__:
        classification = matching__.group(1).strip()
    else:
        classification = 'Unknown'

    matching2 = re.search(r"<reasoning>(.*?)</reasoning>", str(raw_pred), re.DOTALL)
    matching3 = re.search(r"<explanation>\s*([^<]*?)\s*(?:</explanation>|(?=<))", str(raw_pred), re.DOTALL)
    matching4 = re.search(r"<confidence>\s*([^<]*?)\s*(?:</confidence>|(?=<))", str(raw_pred), re.DOTALL)

    if matching2:
        reasoning = matching2.group(1).strip()
    else:
        reasoning = ''

    if matching3:
        explanation = matching3.group(1).strip()
    else:
        explanation = ''

    if matching4:
        try:
            confidence = float(matching4.group(1).strip())
        except:
            confidence = -1
    else:
        confidence = -1

    classification = sanitize_pred_label(classification)

    return classification, reasoning, explanation, confidence

def extract_multi_xml_prediction(raw_pred):

    classification, explanation, confidence = {},  {}, {}

    rule_matches = list(re.finditer(r"<ruleset>\s*([^<]*?)\s*(?:</ruleset>|(?=<))", raw_pred, re.DOTALL | re.IGNORECASE))

    for i, match in enumerate(rule_matches):
        rule_name = match.group(1).strip()
        start = match.end()
        end = rule_matches[i + 1].start() if i + 1 < len(rule_matches) else len(raw_pred)

        block = raw_pred[start:end]

        classification_match = re.search(r"<classification>\s*([^<]*?)\s*(?:</classification>|(?=<))", block, re.DOTALL | re.IGNORECASE)

        if classification_match:
                classification[rule_name] = classification_match.group(1).strip()
        else:
            classification[rule_name] = "Unknown"

        explanation_match = re.search(r"<explanation>\s*([^<]*?)\s*(?:</explanation>|(?=<))", block, re.DOTALL | re.IGNORECASE)

        if explanation_match:
            explanation[rule_name] = explanation_match.group(1).strip()
        else:
            explanation[rule_name] = ""

        confidence_match = re.search(r"<confidence>\s*([^<]*?)\s*(?:</confidence>|(?=<))", block, re.DOTALL | re.IGNORECASE)

        if confidence_match:
            confidence[rule_name] = confidence_match.group(1).strip()
        else:
            confidence[rule_name] = -1

    return classification, explanation, confidence

def sanitize_pred_label(pred_label):

    pred_label = pred_label.replace('Non-Compliant', 'Violated')
    pred_label = pred_label.replace('Compliant', 'Complied')
    pred_label = pred_label.replace('Non-Applicable', 'Not Applicable')

    pred_label = pred_label.strip().title()

    return pred_label if pred_label in ['Complied', 'Violated', 'Not Applicable'] else 'Unknown'

def get_model_outputs(img, template, max_new_tokens = 256):
    messages = [{"role": "user", "content": [{"type": "image"}, {"type": "text", "text": template}]}]
    prompt = processor.apply_chat_template(messages, add_generation_prompt=True, tokenize=False)
    inputs = processor(text=prompt, images=[img], return_tensors="pt").to(model.device)
    prediction = model.generate(
        **inputs,
        max_new_tokens=max_new_tokens,
        do_sample=False,
        # repetition_penalty=1.2,
        pad_token_id=processor.tokenizer.pad_token_id,
        eos_token_id=processor.tokenizer.eos_token_id
    )

    full_pred = processor.decode(prediction[0], skip_special_tokens=False)
    raw_pred = full_pred.split(assistant_token)[-1].replace(processor.tokenizer.eos_token, '').replace(processor.tokenizer.pad_token, '')

    return raw_pred, full_pred


def get_dataset(domain, run_set='all'):

    train_df = pd.read_json(root_dir + f'/experimentation/{domain.lower()}-train.jsonl', lines=True)
    val_df = pd.read_json(root_dir + f'/experimentation/{domain.lower()}-val.jsonl', lines=True)
    test_df = pd.read_json(root_dir + f'/experimentation/{domain.lower()}-test.jsonl', lines=True)

    if run_set == 'all':
        datasets = {'train_df': train_df, 'val_df': val_df, 'test_df': test_df}

    elif run_set == 'train':
        datasets = {'train_df': train_df}

    elif run_set == 'val':
        datasets = {'val_df': val_df}

    elif run_set == 'test':
        datasets = {'test_df': test_df}

    all_unique_images = list(pd.concat(datasets.values())['Image Path'].unique())
    image_cache = {x.replace('data/', ''): load_image(f"{root_dir}{x.replace('data/', '')}") for x in all_unique_images}

    return datasets, image_cache


def get_answers_from_raw_pred(template_id, raw_pred):

      if template_id.startswith('t') and model_name == 'llavacot':

          return extract_xml_prediction(raw_pred)

      if template_id.startswith('t'):

          return sanitize_pred_label(raw_pred), '', '', -1

      elif template_id.startswith('rx'):

          return extract_xml_prediction(raw_pred)

      elif template_id.startswith(('v', 'r')):

          return extract_json_prediction(raw_pred)


# llamavision

def extract_classification_explanation(text):

    classification, explanation, confidence = "Unknown", "", -1

    pattern1 = r"\*\*Classification:\*\*\s*(.*?)\s*(?:\n|$).*?"
    pattern2 = r"\*\*Explanation:\*\*\s*(.*?)\s*(?:\n|$).*?"
    pattern3 = r"\*\*Confidence:\*\*\s*(.*?)\s*(?:\n|$).*?"

    match1 = re.search(pattern1, text, re.DOTALL | re.IGNORECASE)
    if match1:
        classification = match1.group(1).strip()

    match2 = re.search(pattern2, text, re.DOTALL | re.IGNORECASE)
    if match2:
        explanation = match2.group(1).strip()

    match3 = re.search(pattern3, text, re.DOTALL | re.IGNORECASE)
    if match3:
        confidence = match3.group(1).strip()

    return {"classification": classification, "explanation": explanation, "confidence": confidence}
```

```python
# @title
def main(datasets, image_cache, model_name, template_id, domain, rule_template, verbose = False):

    for dfname, df in datasets.items():

        print(f"Running {model_name} with template {template_id} on {dfname} set {domain}")

        start_time = time.time()

        results = []

        if model_name == 'llavacot':
            max_new_tokens = 1024
        elif template_id.startswith(('t', 'v')):
            max_new_tokens = 128
        elif template_id.startswith('r'):
            max_new_tokens = 1024
        elif template_id.startswith('m'):
            max_new_tokens = 2048
        else:
            max_new_tokens = 2048

        # single-rule
        if template_id.startswith(('t', 'v', 'r')):

            for _, row in tqdm(df.iterrows(), total=len(df)):

                domain = row['Domain'].lower()
                filename = row['File']
                image_path = row['Image Path']
                image_id = row['Image ID']
                image_link = row['Image Link']
                rule_key = row['Rule']
                true_label = row['Label']

                # Defaults
                raw_pred = ''
                pred_label = 'Unknown'
                pred_reasoning = ''
                pred_explanation = ''
                pred_confidence = -1

                if not image_path.lower().endswith(('.jpg', '.jpeg', '.png')):
                    continue

                try:
                    img = image_cache[row['Image Path'].replace('data/', '')]
                    template = get_template(template_id, rule_key, rule_template)
                    raw_pred, full_pred = get_model_outputs(img, template, max_new_tokens)

                    try:

                        pred_label, pred_reasoning, pred_explanation, pred_confidence = get_answers_from_raw_pred(template_id, raw_pred)

                    except:

                        result = extract_classification_explanation(raw_pred)
                        pred_label = result.get('classification', 'Unknown')
                        pred_explanation = result.get('explanation', '')
                        pred_confidence = -1

                    # if pred_label == 'Unknown':
                    #     print(raw_pred)
                    #     break

                    results.append({
                        'Domain': domain,
                        'File': filename,
                        'Image ID': image_id,
                        'Image Path': image_path,
                        'Image Link': image_link,
                        'Rule': rule_key,
                        'Label': true_label,
                        'Pred Label': pred_label,
                        'Pred Explanation': pred_explanation,
                        'Pred Confidence': pred_confidence,
                        'Pred Reasoning': pred_reasoning,
                        'Pred Full Response': full_pred,
                    })

                except Exception as e:
                    print(f"\nPrediction failed for {image_path}: {e}")
                    print(raw_pred)
                    results.append({
                        'Domain': domain,
                        'File': filename,
                        'Image ID': image_id,
                        'Image Path': image_path,
                        'Image Link': image_link,
                        'Rule': rule_key,
                        'Label': true_label,
                        'Pred Label': pred_label,
                        'Pred Explanation': pred_explanation,
                        'Pred Confidence': pred_confidence,
                        'Pred Reasoning': pred_reasoning,
                        'Pred Full Response': full_pred,
                    })
                    continue

                if _ == 0:
                    print(template)
                    print('\n--- Sample: ', image_path)
                    print('Rule:', rule_key)
                    print('Label:' , true_label)
                    print('Pred Label:' , pred_label)
                    print('Pred Explanation: ', pred_explanation )
                    print('Pred Confidence: ', pred_confidence )
                    print('Pred Reasoning: ', pred_reasoning)

                    # print('Pred Raw Response: ', raw_pred )

                else:

                    if verbose:
                        print('\n--- Sample: ', image_path)
                        print('Rule:', rule_key)
                        print('Label:' , true_label)
                        print('Pred Label:' , pred_label)
                        # print('Pred Raw Response: ', raw_pred )
                        print('F1 so far', f1_score([x['Label'] for x in results], [x['Pred Label'] for x in results], average="macro"))


        # multi-rule
        elif template_id.startswith('m'):

            df_subset = df[df['File'].apply(lambda x: ' '.join(x.split('_')[:-1])) == df['Rule']]
            refined_rule_names = {k: x.split(':')[0].strip() for k,x in list(custom_rules.items())}

            for _, row in tqdm(df_subset.iterrows(), total=len(df_subset)):

                domain = row['Domain'].lower()
                filename = row['File']
                image_path = row['Image Path']
                image_id = row['Image ID']
                image_link = row['Image Link']
                true_label = row['Label']

                # Defaults
                raw_pred = ''
                pred_label = 'Unknown'
                pred_explanation = ''
                pred_confidence = -1
                pred_reasoning = ''

                if not image_path.lower().endswith(('.jpg', '.jpeg', '.png')):
                    continue

                try:
                    img = image_cache[row['Image Path'].replace('data/', '')]
                    template = get_template(template_id, 'all', rule_template)
                    raw_pred, full_pred = get_model_outputs(img, template, max_new_tokens)


                    if template_id.startswith("mx"):

                        label_dict, explanation_dict, confidence_dict = extract_multi_xml_prediction(raw_pred)

                        for rule_key in custom_rules.keys():

                            pred_label = label_dict.get(f'{refined_rule_names.get(rule_key)}', 'Unknown')
                            pred_label = sanitize_pred_label(pred_label)
                            pred_explanation = explanation_dict.get(f'{rule_key}', '')
                            pred_confidence = confidence_dict.get(f'{rule_key}', -1)

                            results.append({
                                'Domain': domain,
                                'File': filename,
                                'Image ID': image_id,
                                'Image Path': image_path,
                                'Image Link': image_link,
                                'Rule': rule_key,
                                'Label': true_label,
                                'Pred Label': pred_label,
                                'Pred Explanation': pred_explanation,
                                'Pred Confidence': pred_confidence,
                                'Pred Reasoning': pred_reasoning,
                                'Pred Full Response': full_pred,
                            })


                    elif template_id.startswith('m'):

                        full_dict = ast.literal_eval(repair_json(raw_pred))
                        full_dict = flatten_dict(full_dict)

                        for rule_key in custom_rules.keys():
                            pred_label = full_dict.get(f'{refined_rule_names.get(rule_key)}', 'Unknown')
                            pred_label = sanitize_pred_label(pred_label)
                            pred_explanation = full_dict.get('explanation', '')
                            pred_confidence = full_dict.get('confidence', -1)

                            results.append({
                                'Domain': domain,
                                'File': filename,
                                'Image ID': image_id,
                                'Image Path': image_path,
                                'Image Link': image_link,
                                'Rule': rule_key,
                                'Label': true_label,
                                'Pred Label': pred_label,
                                'Pred Explanation': pred_explanation,
                                'Pred Confidence': pred_confidence,
                                'Pred Reasoning': pred_reasoning,
                                'Pred Full Response': full_pred,
                            })


                except Exception as e:
                    print(f"\nPrediction failed for {image_path}: {e}")
                    print(raw_pred)
                    results.append({
                        'Domain': domain,
                        'File': filename,
                        'Image ID': image_id,
                        'Image Path': image_path,
                        'Image Link': image_link,
                        'Rule': rule_key,
                        'Label': true_label,
                        'Pred Label': pred_label,
                        'Pred Explanation': pred_explanation,
                        'Pred Confidence': pred_confidence,
                        'Pred Reasoning': pred_reasoning,
                        'Pred Full Response': full_pred,
                    })
                    continue


        end_time = time.time()

        total_time = end_time - start_time

        mins, secs = divmod(total_time, 60)
        print(f"Total execution time: {int(mins)} min {int(secs)} sec")

        df_results = pd.DataFrame(results)
        df_results.to_csv(f'{save_dir}/{dfname}-{domain}-{model_name}-{template_id}-{rule_template}-{int(mins)}m{int(secs)}s.csv', index = False)

        print('F1 Macro: ', f1_score(df_results['Label'], df_results['Pred Label'], average="macro"))

        valid_labels = ["Complied", "Violated", "Not Applicable"]
        print('Coverage: ', (df_results['Pred Label'].isin(valid_labels)).mean())

        return df_results
```
